K Means Algorithm

Clustering can be defined as the task of identifying subgroups in the data points in the same subgroup are very similar while data points in different clusters are very different.
- Clustering strives to discover homogeneous subgroups within the data so that the data points in each cluster are as comparable as possible based on a similarity measure like Euclidean-based Distance or Correlation-based Distance
- Clustering is Unsupervised learning algorithm.

K Means Clustering
an unsupervised learning technique that seperates items into clusters based on similarities and are distinct to the objects in another cluster.
The letter K denotes the number of clusters you want to create. You must tell the system how many clusters you need beforehand.

How k means works
1. Finding k value
Hit and trail method and elbow method
2. Measuring distance
3. Grouping
4. Repositioning centroids
5. Convergence




```python
import pandas as pd
import numpy as np
from sklearn import datasets
import matplotlib.pyplot as plt
```

```python
data=datasets.load_wine()
```

```python
X=data.data
y=data.target
```

```python
df =pd.DataFrame(X,columns=data.feature_names)
df['Wine Class']=y
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X=sc.fit_transform(X)
```

```python
from sklearn.cluster import KMeans
```

```python
wss=[]
for i in range(1,11):
  kmeans=KMeans(n_clusters = i, init ='k-means++', random_state=0)
  kmeans.fit(X)
  wss.append(kmeans.inertia_)
```

```python
f3,ax =plt.subplots(figsize=(8,6))
plt.plot(range(1,11),wss)
plt.title("The Elbow method")
plt.xlabel("Number of clusters")
plt.ylabel("WSS")
plt.show()
```

```python
N=3
k_means=KMeans(init='k-means++', n_clusters=N)
k_means.fit(X)
labels=k_means.labels_
print(labels)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(labels,y)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(labels,y)
```
