```python
## Value Imputation
# SimpleImputer is a very convenient strategy for missing data imputation, it replaces all missing values with a statistic calculated from the other values in a column.
# Used statistic mean, median, and mode.
```

```python
from sklearn.impute import SimpleImputer
```

```python
print(f"Number of null values before imputing: {df.age.isnull().sum()}")
```

```python
imp = SimpleImputer(strategy='mean')
df['age']= imp.fit_transform(df[['age']])
print(f"Number of null values after imputing: {df.age.isnull().sum()}")
```

```python
def get_parameters(df):
    parameters = {}
    for col in df.columns[df.isnull().any()]:
        if df[col].dtype == "float64" or df[col].dtype == "int64" or df[col].dtype == "int32":
            strategy = 'mean'
        else:
            strategy = 'most_frequent'
        missing_values = df[col][df[col].isnull()].values[0]
        parameters[col] = {'missing_values': missing_values, 'strategy': strategy}
    return parameters

get_parameters(df)
```

```python
parameters = get_parameters(df)
```

```python
for col, param in parameters.items():
  missing_values = param['missing_values']
  strategy = param['strategy']
  imp = SimpleImputer(missing_values=missing_values, strategy=strategy)
  df[col] = imp.fit_transform(df[[col]]).ravel()
# The .ravel() method in NumPy flattens a multi-dimensional array into a 1-dimensional array
```

```python
## Data Scaling
# If the data is any condition has data ponts far from each other, scaling is a technique to make them closer to each other.
```

Scikit learn Normalization Techniques
1. StandardScaler:
Standard Scaler follows the Standard Normal Distribution, i.e. it assumes a normal distribution for data within each feature.
2. MinMaxScaler: For each value in a feature, MinMaxScaler subtracts the minimum value in the feature and then divides by the range
